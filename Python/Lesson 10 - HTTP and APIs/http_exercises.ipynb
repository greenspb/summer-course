### Handle HTTP and Network Errors

**Goal**: Build a small helper function `fetch(url)` that:

- Uses `timeout=3`
- Calls `response.raise_for_status()`
- Catches and prints friendly messages for:
  - `requests.exceptions.Timeout`
  - `requests.exceptions.HTTPError`
  - `requests.exceptions.RequestException`

Test with:

- A valid endpoint: `/posts/1`
- An invalid endpoint: `/not-a-real-route`

✅ *Check*: Valid request succeeds; invalid request prints a clear error without crashing.

---


````python
# Incomplete, ungraded attempt.
def fetch(url):
    response = requests.get(url, timeout=3)
    status = response.raise_for_status()
    if status == 'requests.exceptions.Timeout':
        print('The request timed out.')
    elif status == 'requests.exceptions.HTTPError':
        print('The request returned an HTTP error.')
    elif status == 'requests.exceptions.RequestException':
        print('The request returned a RequestException.')
    return response, status

fetch('https://jsonplaceholder.typicode.com')
  
````

### Exercise 1: Basic Authentication

**Goal**: Use HTTP Basic Auth to call:

- `/basic-auth/student/pass123`

Use username `student` and password `pass123` in your request.

✅ *Check*: Successful call returns status code `200` and JSON showing authentication success.


````python
from requests.auth import HTTPBasicAuth

basic = HTTPBasicAuth("student", "pass123")
baseURL = 'https://httpbin.org'
basic_auth_url = f"{baseURL}/basic-auth/student/pass123"
response = requests.get(basic_auth_url, auth=basic)
print(response)
````

````python
import base64
text_bytes = "username:password".encode("utf-8")
encoded = base64.b64encode(text_bytes)
headers = {"Authorization": "Basic" + encoded.decode("utf-8")}
response = requests.get(f"https://httpbin.org/basic-auth/username/password", headers=headers)
print("Checking basic auth")
print(response.reason, response.status_code)
print(response.text)
print()
````
